```python
import json
```

```python
scale = 1.1

with open('/media/ben/External/repos/midatlantic-idf/public/data/data2/data.json', 'r') as f:
  data = json.load(f)

for scenario in data.values():
  for timeframe in scenario.values():
    for rp in timeframe.values():
      for fips, county in rp.items():
        if fips != 'min' and fips != 'max':
          county['median'] = county['median'] * scale
          county['10%'] = county['10%'] * scale
          county['90%'] = county['90%'] * scale
          county['25%'] = county['25%'] * scale
          county['75%'] = county['75%'] * scale

with open('/media/ben/External/repos/midatlantic-idf/public/data/data2/data.json', 'w') as f:
  json.dump(data, f)
```

```python
scale = 1.1

with open('/media/ben/External/repos/midatlantic-idf/public/data/data2/stations.json', 'r') as f:
  data = json.load(f)

for station, station_dict in data.items():
  for timeframe, timeframe_dict in station_dict.items():
    if timeframe not in ['fips', 'station_name', 'state', 'latitude', 'longitude']:
      for rp_name, rp_value in timeframe_dict.items():
        if isinstance(rp_value, str):
          data[station][timeframe][rp_name] = str(float(rp_value) * scale)
        else:
          for i, val in enumerate(rp_value):
            data[station][timeframe][rp_name][i] = val * scale

with open('/media/ben/External/repos/midatlantic-idf/public/data/data2/stations.json', 'w') as f:
  json.dump(data, f)
```
